```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
from google.colab import files


```

```python
uploaded = files.upload()

# Get uploaded file name
image_name = list(uploaded.keys())[0]

# Load and convert to grayscale
img = Image.open ("/content/runaway.png").convert("L")
img = np.asarray(img) / 255.0  # Normalize to [0, 1]
```

```python
gamma_05 = img ** 0.5
```

```python
gamma_2 = img ** 2
```

```python
r1, r2 = 0.2, 0.8
contrast = np.zeros_like(img)

contrast[img < r1] = 0
contrast[(img >= r1) & (img <= r2)] = (img[(img >= r1) & (img <= r2)] - r1) / (r2 - r1)
contrast[img > r2] = 1
```

```python
plt.figure(figsize=(12, 8))

plt.subplot(2, 2, 1)
plt.imshow(img, cmap='gray')
plt.title("Original Image")
plt.axis('off')

plt.subplot(2, 2, 2)
plt.imshow(gamma_05, cmap='gray')
plt.title("Gamma Correction (γ = 0.5)")
plt.axis('off')

plt.subplot(2, 2, 3)
plt.imshow(gamma_2, cmap='gray')
plt.title("Gamma Correction (γ = 2)")
plt.axis('off')

plt.subplot(2, 2, 4)
plt.imshow(contrast, cmap='gray')
plt.title("Contrast Stretching")
plt.axis('off')

plt.tight_layout()
plt.show()
```
